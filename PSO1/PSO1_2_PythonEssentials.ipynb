## Variables and arithmetic

Python works as a calculator. Variables are created by assignment — no type declarations needed.

```python
V = 4.0      # tank volume, m^3
q = 0.4      # volumetric flow, m^3/min
tau = V / q  # a ratio we will meet again soon (units: min)
print("V/q =", tau, "min")
```

## NumPy arrays

NumPy gives us arrays and vectorized math: operations apply element by element, so we rarely need loops. `np.linspace(a, b, n)` builds `n` evenly spaced points between `a` and `b` — that will be our time axis in every simulation.

```python
import numpy as np

t = np.linspace(0, 50, 501)   # 0 to 50 min, 501 points
y = 1 - np.exp(-t / tau)      # element-wise: exp applied to every entry
print(t[:5])
print(y[:5])
```

## Functions

Wrap reusable computations in functions with `def`. Here is the exponential decay of a tank level toward zero, a curve we will derive from a differential equation in Week 2 of lecture.

```python
def level_decay(t, h0, tau):
    """Liquid level h(t) = h0 * exp(-t/tau)."""
    return h0 * np.exp(-t / tau)

print(level_decay(t=5.0, h0=2.0, tau=10.0))
```

## Plotting with Matplotlib

Every plot you submit this semester needs axis labels **with units**, a legend when there is more than one curve, and a grid. Build that habit now.

```python
import matplotlib.pyplot as plt

h0 = 2.0
plt.figure(figsize=(8, 4))
plt.plot(t, level_decay(t, h0, tau=5.0), label="tau = 5 min")
plt.plot(t, level_decay(t, h0, tau=15.0), label="tau = 15 min")
plt.xlabel("Time (min)")
plt.ylabel("Level h(t) (m)")
plt.title("Exponential decay of tank level")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Your solution here
# 1. time vector

import numpy as np

V = 4.0      # tank volume, m^3
q = 0.4      # volumetric flow, m^3/min
tau = V / q  # a ratio we will meet again soon (units: min)

def level_decay(t, h0, tau):
    """Liquid level h(t) = h0 * exp(-t/tau)."""
    return h0 * np.exp(-t / tau)

t = np.linspace(0, 60, 301)   # 0 to 60 min, 301 points
y = 1 - np.exp(-t / tau)      # element-wise: exp applied to every entry

# 2.-3. plot the three curves with labels, title, legend, grid

import matplotlib.pyplot as plt

h0 = 3.0
plt.figure(figsize=(8, 4))
plt.plot(t, level_decay(t, h0, tau=2.0), label="tau = 2 min")
plt.plot(t, level_decay(t, h0, tau=5.0), label="tau = 5 min")
plt.plot(t, level_decay(t, h0, tau=10.0), label="tau = 10 min")
plt.xlabel("Time (min)")
plt.ylabel("Level h(t) (m)")
plt.title("Exponential decay of tank level")
plt.legend()
plt.grid(True)
plt.show()
```
